# Derivación tensorial del Lagrangiano completo de Lanczos--Lovelock

Este notebook recibe **únicamente un entero**

$
m\ge 0,
$

interpretado como el orden máximo de Lovelock. A partir de él construye **todos** los términos

$
L_0,\ L_1,\ \ldots,\ L_m,
$

y después forma

$
\boxed{
L=\sum_{k=0}^{m}c_kL_k
}.
$

Cada término se construye como

$
L_k^{(D)}=
\frac{1}{16\pi}\frac{1}{2^k}
\delta^{a_1b_1\cdots a_kb_k}_{c_1d_1\cdots c_kd_k}
R^{c_1d_1}{}_{a_1b_1}\cdots R^{c_kd_k}{}_{a_kb_k}.
$

Luego **el Lagrangiano total** pasa por toda la cadena:

$
L\to P^{abcd}\to P^{ab}\to \mathcal R_{ab}
\to \delta(\sqrt{-g}L)
\to \text{Palatini}
\to \text{IBP}
\to E_{ab}.
$

Para órdenes bajos se muestran explícitamente

$
L_0=\frac{1}{16\pi},\qquad
L_1=\frac{R}{16\pi},\qquad
L_2=\frac{1}{16\pi}
\left(R_{abcd}R^{abcd}-4R_{ab}R^{ab}+R^2\right).
$

Para \(k\ge3\), la delta generalizada se mantiene sin expandir para evitar crecimiento factorial.

Primero se construye cada término \(L_k\) por separado. Luego se forma el Lagrangiano final

\[
L=\sum_{k=0}^{m}c_kL_k.
\]

Después se deriva **cada** \(L_k\) respecto del Riemann por regla del producto, y al final se suman las contribuciones:

\[
P^{abcd}=\sum_{k=0}^{m}c_kP_{(k)}^{abcd}.
\]

La divergencia se comprueba término a término usando Bianchi y luego para la suma completa.

\[
